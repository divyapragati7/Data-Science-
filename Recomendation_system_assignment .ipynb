```python
features = df[['genre','rating','members']]
features.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english')

genre_matrix = tfidf.fit_transform(df['genre'])
```

```python
scaler = MinMaxScaler()

numerical_features = scaler.fit_transform(
    df[['rating','members']]
)
```

```python
from scipy.sparse import hstack

combined_features = hstack([
    genre_matrix,
    numerical_features
])
```

```python
cosine_sim = cosine_similarity(combined_features)

print(cosine_sim.shape)
```

```python
indices = pd.Series(df.index,
                    index=df['name']).drop_duplicates()
```

```python
def recommend_anime(title, top_n=10):

    if title not in indices:
        return "Anime not found."

    idx = indices[title]

    similarity_scores = list(
        enumerate(cosine_sim[idx])
    )

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_scores = similarity_scores[1:top_n+1]

    anime_indices = [i[0] for i in similarity_scores]

    recommendations = df[['name','genre','rating']].iloc[
        anime_indices
    ].copy()

    recommendations['Similarity Score'] = [
        i[1] for i in similarity_scores
    ]

    return recommendations
```

```python
recommend_anime("Naruto")
recommend_anime("Death Note")
recommend_anime("One Piece")
```

```python
def recommend_with_threshold(title, threshold=0.5):

    idx = indices[title]

    sim_scores = list(
        enumerate(cosine_sim[idx])
    )

    filtered = [
        score for score in sim_scores
        if score[1] > threshold
    ]

    filtered = sorted(
        filtered,
        key=lambda x:x[1],
        reverse=True
    )

    anime_idx = [i[0] for i in filtered[1:]]

    return df['name'].iloc[anime_idx]
```

```python
recommend_with_threshold("Naruto",0.5)
```

```python
recommend_with_threshold("Naruto",0.7)
```

```python
recommend_with_threshold("Naruto",0.9)
```

Lower threshold gives more recommendations.  
Higher threshold gives fewer but more relevant recommendations.

Difference Between User-Based and Item-Based Collaborative Filtering  
**User-Based Collaborative Filtering: ** Finds users with similar preferences.
Recommends items liked by similar users.  
Example:
If User A and User B like the same anime, anime watched by User B can be recommended to User A.  

**Item-Based Collaborative Filtering **:  Finds items similar to each other.
Recommends items similar to previously liked items.
Example:
If a user likes Naruto, the system recommends anime similar to Naruto.

Comparison
User-Based	Item-Based
User similarity	Item similarity
Dynamic	More stable
Slower for large datasets	Faster for large datasets
Less scalable	More scalable

2. What is Collaborative Filtering?

Collaborative Filtering is a recommendation technique that predicts user preferences based on the behavior of similar users or similar items.

Working Process
Collect user-item interactions.
Measure similarity.
Identify neighbors.
Generate recommendations.
Types
User-Based Collaborative Filtering
Item-Based Collaborative Filtering
Advantages
Personalized recommendations.
No need for item content.
Disadvantages
