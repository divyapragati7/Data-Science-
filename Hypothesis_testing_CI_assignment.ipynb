```python
from scipy.stats import norm
import pandas as pd
import numpy as np
import math
```

```python
#given data
#Sample mean weekly cost: x̄ = 3050
#Sample size: n = 25
#Cost model: W = 1000 + 5X
#Mean units produced: X = 600
#Standard deviation of units: 25
```

```python
x_bar = 3050
X = 600
n = 25
sigma_units = 25
Sigma = 5*sigma_units
alpha = 0.05 #means critical value and xbar is usually less then meu
```

Restaurant owner cclaims that weekly cost is higher than the prediction
Ho_null hypothesis meu = 4000 (when nothing changes)
H1_alternative hypothesis is >4000 (as he claims its higher)

```python
meu  = 1000 +5*X
print("Theoretical Mean Cost (meu): Rs.", meu)
```

```python
sigma = 5 * sigma_units
print("Theoretical Mean Weekly Cost (μ):", meu)
print("Standard Deviation of Weekly Cost (σ):", sigma)
```

```python
print("\nHypotheses:")
print("H0: cost = 4000")
print("H1: cost > 4000")
# ho: x bar greate rthan and equal to meu
#h1: x bar smaller than meu
```

Hypothesis testing we can go with T test Y tesz chi square(categorical) and ANova.

when we go to t test or z test
if sample size is greate than or equal to 30. if not then go for t test but here sample siye is less than 25. so we consider standard deviation sigma and if sigma is given then we go for z test and if it was unknown then we go for t test. so here sigma is given so we go with z test

```python
z = (x_bar- meu) / (sigma / math.sqrt(n))

print("\nZ-Test Statistic:", round(z, 4))
```

```python
critical_value = norm.ppf(1 - alpha)
p_value = 1 - norm.cdf(z)

print("Critical Value (α = 0.05):", round(critical_value, 4))
print("P-value:", round(p_value, 6))
```

```python
if z > critical_value:
    print("\nDecision: Reject H0")
else:
    print("\nDecision: Fail to Reject H0")
```

```python
if z > critical_value:
  print('Conclusion: there is a strong eveidence that the weekly operating costs are higher than the model suggests')
else:
  print('conclusion: there is not enough evidene to support the claim that weekly operating costs are higher than model suggests')
```

Background  
In quality control processes, especially when dealing with high-value items, destructive sampling is a necessary but costly method to ensure product quality. The test to determine whether an item meets the quality standards destroys the item, leading to the requirement of small sample sizes due to cost constraints.
Scenario  
A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.
Data  
A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:  
1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29
